#Tuple
##ordered collection of items ()
## One main difference is that tuple is immutable
# var_name=list[[stu1],[stu2]...]

```python
#create a tuple
stu=(101,"Sanjiv","CSE",8.9)
print(stu)

#tuples using index
print(stu[1])

#negative index
print(stu[-2])

subj=("DSA","DSV","OS","CN","Eng","Maths")
print(len(subj))

#slicing
print(subj[1:5])
print(subj[:5])
print(subj[4:])

#nested tuple
tup=(101,"Rahul",(85,98,76))
print(tup[2][1])

#count function
marks=[75,78,68,75,89,75,92,98]
print("Time of occurrence:",marks.count(75))

#element is present or not
print("python" in subj)
```

# SET
## unordered collection of items
## mutable
##it doesn't support for index
## it gives output with unique values
##doesn't depend on order every time u run order will be diff

```python
#symbol is{}
#create a set with similar items
marks={75,88,68,78,75,55,68,99}
print(marks)

#set doesn't support the index
print(marks[1])
```

```python
#creat a emepty set
a={}
a=set()
#check its type
print(type(a))
```

```python
#add an item in set
stu={"Amit","Kiran","Sneha"}
stu.add("Anshu")
print(stu)

#remove an item from the set
stu.remove("Kiran")
print(stu)

stu.clear()
print(stu)
```

```python
#on deleting whole set will be deleted
marks={78,90,88,69}
del marks
print(marks)
```

```python
from enum import unique
#set operations
python_stud={"Anshu","Ayushi","Aniket","Raj"}
java_stud={"Anshu","karan","Nishank","Raj"}

#for union
result=python_stud.union(java_stud)
print("Union Operation:",result)

#for intersectionoperation by using set
inter = python_stud.intersection(java_stud)
print("Intesection Operation:",inter)

#to find difference with to set
res_diff=python_stud.intersection(java_stud)
print("Difference of two set: ",res_diff)

#convert list to set
stud_id=[101,102,103,102,103,104,105]
unique_stud_id=set(stud_id)
print(unique_stud_id)
```
